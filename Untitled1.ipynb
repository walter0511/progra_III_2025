EJERCICIO en python realizar la clasificacion de edades de los alumnos cuantos son mayores de edad y cuantos son menores de edad.

```python
import matplotlib.pyplot as plt
import seaborn as sns




```

```python
edad['estado'] = edad['edad'].apply(lambda n: 'mayor de edad' if n >= 18 else 'menor de edad')
```

```python
sns.set(style="whitegrid")
sns.countplot(data=edad, x='estado', hue='estado', palette={'menor de edad': 'green', 'mayor de edad': 'red'}, legend=False)
plt.title('Alumnos mayores o menor de edad')
plt.xlabel('Estado')
plt.ylabel('Cantidad de alumnos')
plt.show()
```

Entrenar una IA con lalibreria de tensorflow


```python
import tensorflow as tf
import pandas as pd
import numpy as np
```

```python
grados = pd.read_csv("/content/kelvin.csv", sep=";", encoding="utf-8")
print(grados, grados["k"])
```

```python
c = grados["c"]
k = grados["k"]

print(c,k)
```

```python
modelo = tf.keras.Sequential()
modelo.add(tf.keras.layers.Dense(units=1, input_shape=[1]))
```

```python
modelo.compile(optimizer=tf.keras.optimizers.Adam(0.01), loss='mean_squared_error')
```

```python
epocas = modelo.fit(c, k, epochs=690, verbose=1)
```

```python
g = int(input("Celsius: "))
resp = modelo.predict(np.array([g]))
print("kelvin: ", resp[0][0])
```
